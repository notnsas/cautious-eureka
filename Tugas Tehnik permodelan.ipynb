```python
# Kolom yang ingin dipertahankan
keep_cols = ['amount', 'location', 'is_fraud']

# Drop semua kolom selain keep_cols untuk setiap df
dfs = [df1[keep_cols], df2[keep_cols], df3[keep_cols]]

# Gabungkan semuanya
df = pd.concat(dfs, ignore_index=True)
```

```python
from datetime import date
from uuid import UUID, uuid4
from enum import Enum
from pydantic import BaseModel
from pandantic import Pandantic

# Membuat class fraud untuk validasi pake pydantics
class Fraud(BaseModel):
    amount: float
    location: str
    is_fraud: int

validator = Pandantic(schema=Fraud)

# DataFrame untuk mengecek apakah invalid
df_invalid = df
# Pengecekan
df_valid = validator.validate(dataframe=df_invalid, errors="skip")

```

```python
# Drop nilai kosong
df_valid = df_valid.dropna()
```

```python
# --- Imports ---
import pandas as pd
import numpy as np
from imblearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from imblearn.over_sampling import SMOTE
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from imblearn.ensemble import BalancedRandomForestClassifier, RUSBoostClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.impute import SimpleImputer
from joblib import dump
from imblearn.under_sampling import RandomUnderSampler
from imblearn.ensemble import BalancedBaggingClassifier
from sklearn.tree import DecisionTreeClassifier

# Memisah X dan y
X = df_valid.drop('is_fraud', axis=1)
y = df_valid['is_fraud']

# Train test split
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.8, test_size=0.2,
                                                      random_state=0, stratify=y)
numerical_cols = X.select_dtypes(include=np.number).columns
categorical_cols = X.select_dtypes(exclude=np.number).columns

# Numerik pipeline: imputer (mean) + scaler
numerical_pipeline = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='mean')),
    ('scaler', StandardScaler())
])

# Kategorikal pipeline: imputer (paling sering) + one-hot encoder
categorical_pipeline = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])

# Gabungkan keduanya ke column transformer
preprocessor = ColumnTransformer(transformers=[
    ('num', numerical_pipeline, numerical_cols),
    ('cat', categorical_pipeline, categorical_cols)
])

# smt = SMOTE(random_state=42)
over = SMOTE(sampling_strategy=0.1)
# over = SMOTE()
under = RandomUnderSampler(sampling_strategy=0.5)

# --- Models ---
models = {
    "Random Forest Classifier": RandomForestClassifier(class_weight='balanced_subsample', random_state=42),
    "RUSBoost Classifier": RUSBoostClassifier(random_state=42),
    "Logistic Regression": LogisticRegression(max_iter=1000, random_state=42),
    "Gradient Boosting Classifier": GradientBoostingClassifier(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=5,
    min_samples_leaf=20,
    subsample=0.8,
    random_state=42
),
    "Balanced Random Forest Classifier": BalancedRandomForestClassifier(random_state=42),
}

# --- Train, evaluasi, dan simpan models ---
for name, model in models.items():
    pipeline = Pipeline(steps=[
        ('preprocessor', preprocessor),
        ('over', over),
        ('under', under),
        ('model', model)
    ])

    # Train model
    pipeline.fit(X_train, y_train)

    # Prediksi di test set
    y_pred = pipeline.predict(X_test)

    # Kalkulasi metriks
    acc = accuracy_score(y_test, y_pred)

    # Print hasil
    print(f"\n📊 {name}")
    print(f"Accuracy : {acc:.3f}")

    # Simpan model
    filename = f"{name.replace(' ', '_').lower()}.joblib"
    dump(pipeline, filename)
    print(f"Model ke simpan ke {filename}")
```
